```python
x = np.array([0, 1, 2])

y = np.array([0, 1])
coords = np.array(np.meshgrid(x,y))
coords
```

```python
coords.shape
```

```python
coords=np.append(coords,np.array([np.zeros(shape=(2,3))]),0)
coords
```

```python
pointcds=np.reshape(coords, (3,6)).T
pointcds
```

```python
vtkarr = vtk.vtkDoubleArray()
```

```python
pointcds.size
```

```python
pointcds.shape
```

```python
pointcds.copy(order='C').flags
```

```python
vtkarr.SetNumberOfComponents(pointcds.shape[1])
vtkarr.SetNumberOfTuples(pointcds.shape[0])
vtkarr.SetVoidArray(pointcds.copy(order='C'), pointcds.size, 1)
```

```python
points = vtk.vtkPoints()
```

```python
points.SetData(vtkarr)
```
