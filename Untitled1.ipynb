```python
from scipy import sparse as ssp
```

```python
import numpy as np
```

```python
data = np.array([[1, 2, 3, 4]]).repeat(2, axis=0)
print(data)

offsets = np.array([0, -1])

ssp.dia_array((data, offsets), shape=(4, 4)).toarray()
```

```python
tau = 0.1
h = 0.1
n = 10
m = 2
N = n*m
ex = np.ones(N)
offsets = np.array([-n, -1,0,1,n])
ipos = np.repeat(np.array([0] + [(i + 0.5)/(2*h*h*(i)*(i)) for i in range(1,n-1)] + [0]),m)
print(ipos)
ineg = np.repeat(np.array([0] + [(i + 0.5)/(2*h*h*(i)*(i)) for i in range(1,n-1)] + [0]),m)
print(ineg)
data = np.array([-ex*1/(2*h*h), -ineg, ex*(1/tau + 2/(h*h)), -ipos, -ex*1/(2*h*h) ],dtype='float64')
print(data.shape)
print(ex.shape)
print(ineg.shape)
print(ipos.shape)
print(data)
print
A = ssp.dia_array((data, offsets), shape=(N, N))

```

```python
ssp.linalg.inv(A.asformat('csr'))
```
