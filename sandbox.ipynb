```python

import numpy as np
from scipy.integrate import solve_ivp

# Define the kinetic-diffusion equation
def fun(t, y, D, k, dx):
    c = y
    N = len(c)
    c_left = c[0]
    c_right = c[N-1]
    f = np.zeros(N)
    f[0] = (D/(dx**2))*(c[1]-2*c[0])-(k/2)*c[0]
    f[N-1] = (D/(dx**2))*(c[N-2]-2*c[N-1])-(k/2)*c[N-1]
    for i in range(1,N-1):
        f[i] = (D/(2*dx**2))*(c[i+1]-2*c[i]+c[i-1])-(k/2)*c[i]
    return f

# Define the initial conditions
def initial_condition(x):
    return np.sin(np.pi*x/5)

# Define the boundary conditions
def boundary_conditions(c):
    return [c[0]-1, c[-1]-0]

# Define the parameters
D = 0.1
k = 0.01
L = 10
dx = 0.1
x = np.arange(0, L+dx, dx)
t_span = (0, 100)
c0 = initial_condition(x)

# Solve the equation
sol = solve_ivp(fun, t_span, c0, method='RK45', args=(D, k, dx), jac=None, events=None, rtol=1e-6, atol=1e-6)

```
